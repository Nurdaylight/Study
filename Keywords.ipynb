```python
import re
import spacy
from wordfreq import zipf_frequency
from collections import Counter

nlp = spacy.load("en_core_web_sm")

# ---- Tuning knobs ----
ZIPF_MIN = 3.0          # lower => rarer words allowed
ZIPF_MAX = 6.7          # upper => very common function words filtered
MIN_COUNT = 2           # phrase must appear at least N times in text
MAX_PHRASE_LEN = 4      # keep phrases short for learning

# Common “junk heads” (optional, expand as needed)
JUNK_HEADS = {
    "way","thing","place","time","fact","instance","part","means","kind","sort",
    "moment","story","version","detail","appeal"
}

def is_content_token(t):
    """Keep content POS; drop pronouns/determiners/etc."""
    return t.pos_ in {"NOUN", "PROPN", "VERB", "ADJ", "ADV"}

def is_bad_token(t):
    """Hard exclusions."""
    if t.is_stop:
        return True
    if t.pos_ in {"PRON", "DET", "AUX", "ADP", "CCONJ", "SCONJ", "PART", "NUM"}:
        return True
    if t.ent_type_ and t.ent_type_ in {"PERSON", "GPE", "LOC", "ORG", "NORP", "FAC"}:
        return True  # drop named entities
    return False

def zipf_ok(word):
    f = zipf_frequency(word.lower(), "en")
    return ZIPF_MIN <= f <= ZIPF_MAX

def normalize_phrase(tokens):
    # lemmatize verbs/nouns; keep adjectives as surface
    out = []
    for t in tokens:
        if t.pos_ in {"NOUN","VERB"}:
            out.append(t.lemma_.lower())
        else:
            out.append(t.text.lower())
    return " ".join(out)

def valid_phrase(doc_tokens):
    # length
    if not (2 <= len(doc_tokens) <= MAX_PHRASE_LEN):
        return False

    # no punctuation / spaces
    if any(t.is_punct or t.is_space for t in doc_tokens):
        return False

    # no stopword at boundaries
    if doc_tokens[0].is_stop or doc_tokens[-1].is_stop:
        return False

    # reject if any token is “bad” (pronoun, det, entity, etc.)
    if any(is_bad_token(t) for t in doc_tokens):
        return False

    # require at least one noun or verb (anchored meaning)
    if not any(t.pos_ in {"NOUN","VERB"} for t in doc_tokens):
        return False

    # frequency band: all content words must be in learnable range
    for t in doc_tokens:
        if t.pos_ in {"NOUN","VERB","ADJ","ADV"}:
            if not t.is_alpha:
                return False
            if not zipf_ok(t.text):
                return False

    # optional: remove phrases whose syntactic head is junk
    # approximate head = last NOUN/VERB in phrase
    head = next((t.lemma_.lower() for t in reversed(doc_tokens) if t.pos_ in {"NOUN","VERB"}), "")
    if head in JUNK_HEADS:
        return False

    return True

def extract_candidates(doc):
    cand = []

    # 1) Phrasal verbs: VERB + particle (get up, run out, carry out)
    for t in doc:
        if t.pos_ == "VERB":
            prt = [c for c in t.children if c.dep_ == "prt"]
            for p in prt:
                phrase_tokens = [t, p]
                if valid_phrase(phrase_tokens):
                    cand.append(normalize_phrase(phrase_tokens))

    # 2) Pattern-based bigrams/trigrams (collocation-like)
    #   ADJ+NOUN, NOUN+NOUN, VERB+NOUN, ADV+ADJ
    tokens = [t for t in doc if t.is_alpha and not t.is_space]
    for i in range(len(tokens) - 1):
        a, b = tokens[i], tokens[i+1]
        pos2 = (a.pos_, b.pos_)
        if pos2 in {("ADJ","NOUN"), ("NOUN","NOUN"), ("VERB","NOUN"), ("ADV","ADJ")}:
            phrase = [a, b]
            if valid_phrase(phrase):
                cand.append(normalize_phrase(phrase))

    # 3) Short noun chunks, but stricter than spaCy defaults
    for chunk in doc.noun_chunks:
        chunk_tokens = [t for t in chunk if t.is_alpha]
        # drop leading determiners even if not in chunk tokens
        chunk_tokens = [t for t in chunk_tokens if t.pos_ != "DET"]
        if valid_phrase(chunk_tokens):
            cand.append(normalize_phrase(chunk_tokens))

    return cand

def extract_learning_vocab(text):
    text = re.sub(r"\s+", " ", text).strip()
    doc = nlp(text)

    cands = extract_candidates(doc)
    counts = Counter(cands)

    # enforce repetition in the text (removes many one-off weird phrases)
    kept = [p for p, c in counts.items() if c >= MIN_COUNT]

    # sort: highest count first, then length, then alphabetic
    kept.sort(key=lambda p: (-counts[p], len(p.split()), p))
    return kept

# ---- Example ----
```

```python
from simalign import SentenceAligner

src = "She carried out the experiment and came up with a new idea."
tgt = "Она провела эксперимент и придумала новую идею."

# 'a' stands for ArgMax (typically Intersection), 'm' for MwMF, 'i' for IterMax
aligner = SentenceAligner(model="xlmr", token_type="bpe", matching_methods=["a"])
aligned = aligner.get_word_aligns(src, tgt)

# The library returns the key "inter" for method "a"
print(aligned)
```

```python
# Split sentences into words to match the aligner's indices
src_tokens = src.split()
tgt_tokens = tgt.split()

# Get the list of index pairs from the 'inter' (Intersection) method
pairs_indices = aligned['inter']

# Convert indices to words
text_pairs = [(src_tokens[i], tgt_tokens[j]) for i, j in pairs_indices]

# Print the results
for s, t in text_pairs:
    print(f"{s} <-> {t}")
```

```python
import re

def split_sentences(text: str):
    text = re.sub(r"\s+", " ", (text or "")).strip()
    if not text:
        return []
    # Split on end punctuation followed by whitespace (no A-Z assumption)
    sents = re.split(r"(?<=[.!?])\s+", text)
    # If still one giant sentence (no punctuation), fallback to chunking
    return [s.strip() for s in sents if s.strip()]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def monotonic_sentence_align(src_sents, tgt_sents, window=8):
    """
    Monotonic alignment: each src sentence aligns to 1 tgt sentence, in order.
    Uses local cosine similarity + Viterbi-style DP with a band (window).
    """
    vec = TfidfVectorizer(ngram_range=(1,2), min_df=1)
    all_sents = src_sents + tgt_sents
    X = vec.fit_transform(all_sents)
    Xs = X[:len(src_sents)]
    Xt = X[len(src_sents):]

    # Precompute similarity band
    S = np.full((len(src_sents), len(tgt_sents)), -1e9, dtype=np.float32)
    for i in range(len(src_sents)):
        j0 = max(0, i - window)
        j1 = min(len(tgt_sents), i + window + 1)
        sims = cosine_similarity(Xs[i], Xt[j0:j1]).ravel()
        S[i, j0:j1] = sims

    # DP
    dp = np.full_like(S, -1e9)
    back = np.full(S.shape, -1, dtype=np.int32)

    dp[0] = S[0]
    for i in range(1, len(src_sents)):
        for j in range(len(tgt_sents)):
            # enforce monotonicity: previous j' <= j
            best_jp = np.argmax(dp[i-1, :j+1])
            dp[i, j] = dp[i-1, best_jp] + S[i, j]
            back[i, j] = best_jp

    # backtrack best end
    j = int(np.argmax(dp[-1]))
    pairs = []
    for i in range(len(src_sents)-1, -1, -1):
        pairs.append((i, j))
        j = int(back[i, j]) if i > 0 else j
    return list(reversed(pairs))
```

```python
import re
from googletrans import Translator

translator = Translator()

def _pick_anchor(s: str, max_len: int = 40) -> str:
    s = s.strip()
    return s[:max_len] if len(s) > max_len else s

async def translate_chunk_contextual(text: str, chunk: str, *, anchor_len: int = 30) -> str:
    # 1) Locate first occurrence of chunk
    m = re.search(re.escape(chunk), text)
    if not m:
        return text

    start, end = m.span()

    # 2) Build a local window around the chunk (stable indexing in EN)
    w_start = max(0, start - anchor_len)
    w_end = min(len(text), end + anchor_len)
    window = text[w_start:w_end]

    # Split window using known indices relative to window
    rel_start = start - w_start
    rel_end = end - w_start
    left = window[:rel_start]
    mid = window[rel_start:rel_end]   # == chunk
    right = window[rel_end:]

    # Optional: shorten anchors a bit to improve match robustness
    left_anchor = _pick_anchor(left)
    right_anchor = _pick_anchor(right)

    # 3) Translate the window (context-preserving for the chunk)
    ru_window = (await translator.translate(window, src="en", dest="ru")).text

    # 4) Translate anchors to find boundaries inside ru_window
    ru_left = (await translator.translate(left_anchor, src="en", dest="ru")).text if left_anchor else ""
    ru_right = (await translator.translate(right_anchor, src="en", dest="ru")).text if right_anchor else ""

    # 5) Find ru_left and ru_right in ru_window to extract only the middle
    lo = 0
    if ru_left:
        i = ru_window.find(ru_left)
        if i != -1:
            lo = i + len(ru_left)

    hi = len(ru_window)
    if ru_right:
        j = ru_window.find(ru_right, lo)
        if j != -1:
            hi = j

    ru_mid = ru_window[lo:hi].strip()

    # If anchors fail (common with MT), fall back to translating just the chunk
    if not ru_mid or (ru_left and ru_left not in ru_window) or (ru_right and ru_right not in ru_window):
        ru_mid = (await translator.translate(mid, src="en", dest="ru")).text

    # 6) Replace only the first occurrence in the original text
    return text[:start] + ru_mid + text[end:]


# example
```
