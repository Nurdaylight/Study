```python
POSITIVE = ' Positive'
NEGATIVE = ' Negative'
UNKNOWN = ' Unknown/Neutral'
BOTH = ' both'

POSITIVE_WORDS = set (['determine', 'success', 'cooperate', 'appreciate', 'confidence', 'engage',
                       'effectively', 'good', 'win', 'respect', 'capable', 'friend', 'appreciation',
                       'ally', 'concert', 'unity', 'ready', 'bolster', 'comprehensive', 'democratic', 'readiness', 
                       'independence', 'support', 'advance', 'incredible', 'emotional', 'courageous', 'brave', 'good', 
                       'recover', 'thanks', 'safe', 'famous''victory', 'triumph', 'train', 'equip', 'boost', 'support', 'success', 
                       'defend', 'deliver', 'strong', 'aid', 'progress', 'volunteer', 'counter-assault', 'ally']
)

NEGATIVE_WORDS = set (['disturb', 'vulnerable', 'hamper', 'fear', 'manipulate', 'autocratic', 'arrest', 
                       'embarrass', 'leak', 'enemy', 'strike', 'invade', 'debris', 'threat', 'concern', 
                       'espionage', 'disinformation''conflict', 'threat', 'invade', 'disrupt', 'embolden', 'pressure', 'war', 'aggressive', 
                       'risk', 'conflict', 'concern', 'fail', 'against', 'terrible', 'difficult', 'injury', 'bomb', 'destroy', 'war', 'nervous', 
                       'scare', 'disorient', 'attack', 'frontline', 'casualty', 'danger','punish', 'casualty', 'injure', 'fail', 'frustration',
                       'drop', 'lack', 'occupy', 'fight', 'counter-offensive', 'worry', 'frontline', 'imminent', 'toll'])

def posneg(words):
    pwlen = len(POSITIVE_WORDS.intersection(words))
    nwlen = len(NEGATIVE_WORDS.intersection(words))
    
    if pwlen > 0 and nwlen == 0:
        return POSITIVE
    elif pwlen == 0 and nwlen > 0:
        return NEGATIVE
    elif pwlen > 0 and nwlen > 0:
        return BOTH
    else:
        return UNKNOWN 
    
def count_posneg(sentences):
    sents = Counter()
    words = Counter()
    
    for sentence in sentences:
        gender = posneg(sentence) # 'male', 'female', 'unknown', 'both'
        sents[gender] += 1 #the same as sents[gender] = sents[gender] + 1
        words[gender] += len(sentence)
        
    return sents, words 

def parse_posneg(text):
    # process the input text file into sentences, which are further processed into 
    #tokenized words
    sentences = [
        [word.lower() for word in nltk.word_tokenize(sentence)]
        for sentence in nltk.sent_tokenize(text)
    ]
    
    sents, words = count_posneg(sentences)
    total = sum(words.values())
    
    for gender, count in words.items():
        pcent = (count/total)*100 
        nsents = sents[gender]
        
        print(
            "{:0.3f}%{} ({} sentences)".format(pcent,gender,nsents)
    
        )
    
```

```python
with open('taiwan1.txt', 'r', encoding = 'UTF8') as file:
    doc = file.read()
parse_posneg(doc)
    
```
